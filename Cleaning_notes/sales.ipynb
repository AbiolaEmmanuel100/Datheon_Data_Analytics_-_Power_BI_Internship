```python
import pandas as pd

sales = pd.read_csv('../superstore_sales.csv', skiprows= 0, nrows =9995)
region = pd.read_csv('../superstore_sales.csv', skiprows= 9995, nrows=4)
goods_returned= pd.read_csv('../superstore_sales.csv',skiprows=10000,)

# Columns Cleaning
sales.columns=sales.columns.str.strip().str.lower().str.replace(' ', '_')


```

```python
print("SALES")
display(sales.head())
print(sales.shape)
```

```python
# DROPPED COLUMNS
sales= sales.drop(columns=['row_id'])
```

```python
print(f'SALES:  {sales.columns.to_list()}')
```

```python
# CONVERT SALES DATES DATA TYPE
date_cols = ["order_date", "ship_date"]

for col in date_cols:
    sales[col] = pd.to_datetime(sales[col], errors="coerce")
display(sales.info())
```

```python
sales = sales.apply(
    lambda col: col.str.strip().str.replace(r"\s+", " ", regex=True)
    if pd.api.types.is_string_dtype(col)
    else col
)
sales["order_id"] = (
    sales["order_id"]
    .str.replace(r"\s*-\s*", "-", regex=True)
    .str.replace(r"\s+", "-", regex=True)
)
```

```python
display(sales.loc[sales.duplicated(keep=False)])
sales=sales.drop_duplicates()
```

```python
sales.to_csv('../sales_cleaned.csv', index=False)
```
