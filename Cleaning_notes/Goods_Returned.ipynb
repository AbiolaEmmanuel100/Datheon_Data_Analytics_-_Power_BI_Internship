```python
import pandas as pd

sales = pd.read_csv('../superstore_sales.csv', skiprows= 0, nrows =9995)
region = pd.read_csv('../superstore_sales.csv', skiprows= 9995, nrows=4)
goods_returned= pd.read_csv('../superstore_sales.csv',skiprows=10000,)


# Columns Cleaning
goods_returned.columns=goods_returned.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
print("GOODS RETURNED")
display(goods_returned.head())
print(goods_returned.shape)
```

```python
goods_returned['returned']=goods_returned['returned'].str.strip().str.capitalize()
goods_returned["order_id"] = (

    goods_returned["order_id"]
    .str.strip()
    .str.replace(r"\s*-\s*", "-", regex=True)
    .str.replace(r"\s+", "-", regex=True)
)
```
