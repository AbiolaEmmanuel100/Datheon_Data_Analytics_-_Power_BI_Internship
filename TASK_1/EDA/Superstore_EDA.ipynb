```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


sales = pd.read_csv('sales_cleaned.csv')
region = pd.read_csv('region_cleaned.csv')
goods_returned= pd.read_csv('goods_returned_cleaned.csv')
```

```python
# Year_Month Extraction
sales["year_month"] = pd.to_datetime(sales["order_date"], errors="coerce").dt.to_period("M")
sales['year_month']
```

```python
# SALES AND PROFIT BY CATEGORY AGGREGATION

Sales_profit_by_category = sales.groupby('category').agg(
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum')

) 
Sales_profit_by_category.round(2).reset_index()
```

```python
# TOTAL SALES BY CATEGORY

sns.barplot(
    data=Sales_profit_by_category.reset_index(),
    x="category",
    y="total_sales"
)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
# TOTAL PROFIT BY CATEGORY

sns.barplot(
    data=Sales_profit_by_category.reset_index(),
    x="category",
    y="total_profit"
)

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.show()
```

```python
# SALES AND PROFIT BY REGION AGGREGATION

Sales_profit_by_Region=sales.groupby('region').agg(total_sale=('sales','sum'),total_profit=('profit','sum'))

Sales_profit_by_Region.round(2).reset_index()
```

```python
# TOTAL SALES BY REGION
plt.figure(figsize=(10, 5))

sns.barplot(
    data=Sales_profit_by_Region.reset_index(),
    x="region",
    y="total_sale"
)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

```python
# TOTAL PROFIT BY REGION
plt.figure(figsize=(10, 5))

sns.barplot(
    data=Sales_profit_by_Region.reset_index(),
    x="region",
    y="total_profit"
)

plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.show()
```

```python
# SALES AND PROFIT BY CUSTOMER SEGMENT AGGREGATION

Sales_profit_by_Customer_Segement=sales.groupby('segment').agg(total_sale=('sales','sum'),total_profit=('profit','sum'))

Sales_profit_by_Customer_Segement.round(2).reset_index()
```

```python
# TOTAL SALES BY CUSTOMERS SEGMENT

plt.figure(figsize=(10, 5))
plt.figure(figsize=(10, 5))

sns.barplot(
    data=Sales_profit_by_Customer_Segement.reset_index(),
    x="segment",
    y="total_sale"
)

plt.title("Total Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.show()
```

```python
# TOTAL PROFIT BY CUSTOMERS SEGMENT

plt.figure(figsize=(10, 5))
plt.figure(figsize=(10, 5))

sns.barplot(
    data=Sales_profit_by_Customer_Segement.reset_index(),
    x="segment",
    y="total_profit"
)

plt.title("Total Profit by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Profit")
plt.show()
```

```python
monthly_sales = sales.groupby("year_month").agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum")
    ).reset_index()

monthly_sales["year_month"] = (
    monthly_sales["year_month"].astype(str)
)
```

```python
# MONTHLY SALES VELOCITY

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_sales,
    x="year_month",
    y="total_sales",
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# MONTHLY PROFIT VELOCITY

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_sales,
    x="year_month",
    y="total_profit",
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# DISCOUNT VS PROFIT

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=sales,
    x="discount",
    y="profit",
    alpha=0.5
)

plt.title("Discount vs Profit")
plt.xlabel("Discount Rate")
plt.ylabel("Profit")
plt.axhline(y=0, linestyle="--")
plt.show()
```
